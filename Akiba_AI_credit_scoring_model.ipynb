#**Akiba AI Credit Risk Scoring for African SACCOs**

##**Project Overview**

AkibaAI is an offline-based XGBoost classifier model for credit risk scoring using mobile money transaction data. The system is designed for frontline loan officers at rural and semi-urban Savings and Credit Co-
operative Organizations (SACCOs). The system ingests customer-consented mobile transaction logs, executes an XGBoost machine learning model, and
outputs transparent credit risk ratings. Operating on local hardware without cloud dependencies, it bridges the financial inclusion gap while respecting severe regional infrastructure and data privacy
constraints


## **Objectives**


*  Ingestion: Build an offline parser for regional SMS logs.
*  ML/XAI: Deploy 8-bit quantized XGBoost
via ONNX (< 2s inference) with localized
SHAP risk narratives
* Security: Implement zero-PII local storage
via AES-256 SQLCipher to satisfy regional
data protection acts
